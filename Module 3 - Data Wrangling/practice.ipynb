```python
import pandas as pd

# Table of employees
df1 = pd.DataFrame({
    'emp_id': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie']
})

# Table of salaries
df2 = pd.DataFrame({
    'emp_id': [2, 3, 4],
    'salary': [60000, 70000, 80000]
})
```

```python
merged = pd.merge(df1, df2, on='emp_id')
```

```python
merged.head()
```

```python
merged_outer = pd.merge(df1, df2, on='emp_id', how='outer')
```
