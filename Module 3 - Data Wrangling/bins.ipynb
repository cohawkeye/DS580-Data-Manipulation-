```python
import numpy as np
import pandas as pd
```

```python
#Create a simple Dataframe going from 0 to 29
data = pd.DataFrame({'values':range(0,30)})
```

```python
data
```

```python
#bins 0-10;11-25;>25
bins = [0,10,25,100]

# These will create the bins:
# (0,10) - with the low part not Inclusive; high part inclusive; going from 1 to 10
# (10,25) - with the low part not Inclusive; high part inclusive; going from 11 to 25
# (25,100) - with the low part not Inclusive; high part inclusive; going from 26 to 99

data['my_bins'] = pd.cut(data['values'],bins)
data
```

```python
#bins 0-10;11-25;>25
bins = [-1,10,25,100]

# These will create the bins:
# (-1,10) - with the low part not Inclusive; high part inclusive; going from 0 to 10
# (10,25) - with the low part not Inclusive; high part inclusive; going from 11 to 25
# (25,100) - with the low part not Inclusive; high part inclusive; going from 26 to 99

data['my_bins'] = pd.cut(data['values'],bins,)
data
```

```python
#bins 0-10;11-25;>25
bins = [0,10,25,100]

# NOTE THE PARAMETER include_lowest=True; These will create the bins:
# (-0.0001,10) - with the low not Inclusive; high part inclusive; going from 0 to 10
# (10,25) - with the low part not Inclusive; high part inclusive; going from 11 to 25
# (20,100) - with the low part not Inclusive; high part inclusive; going from 26 to inf

data['my_bins'] = pd.cut(data['values'],bins, include_lowest=True)
data
```

```python
#bins 0-10;11-25;>25
bins = [np.NINF ,10,25,np.inf]

# NOTE THE VALUES np.NINF (negative infinity) and the np.inf (positive infinity); These will create the bins:
# (-inf,10) - with the low part not Inclusive; high part inclusive; going from -inf to 10
# (10,25) - with the low part not Inclusive; high part inclusive; going from 11 to 25
# (20,inf) - with the low part not Inclusive; high part inclusive; going from 26 to inf

data['my_bins'] = pd.cut(data['values'],bins)
data
```

```python
#bins 0-10;11-25;>25
bins = [np.NINF ,10,25,np.inf]

# NOTE THE VALUES np.NINF (negative infinity) and the np.inf (positive infinity); These will create the bins:
# (-inf,10) - with the low part not Inclusive; high part inclusive; going from -inf to 10
# (10,25) - with the low part not Inclusive; high part inclusive; going from 11 to 25
# (20,inf) - with the low part not Inclusive; high part inclusive; going from 26 to inf

data['my_bins'] = pd.cut(data['values'],bins, labels=['0-10','11-25','>25'])
data
```

```python
#bins 0-10;11-25;>25
bins = [0,11,26,np.inf]

# NOTE THE right=False; This makes the low part INCLUSIVE and the High part NOT INCLUSIVE
# (0,11) - with the low part INCLUSIVE; high part NOT inclusive; going from 0 to 10
# (11,26) - with the low part INCLUSIVE; high part NOT inclusive; going from 11 to 25
# (26,inf) - with the low part INCLUSIVE; high part NOT inclusive; going from 26 to inf

data['my_bins'] = pd.cut(data['values'],bins, right=False)
data
```

```python
#bins 0-10;11-25;>25
bins = [0,11,26,np.inf]

# NOTE THE right=False; This makes the low part INCLUSIVE and the High part NOT INCLUSIVE
# (0,11) - with the low part INCLUSIVE; high part NOT inclusive; going from 0 to 10
# (11,26) - with the low part INCLUSIVE; high part NOT inclusive; going from 11 to 25
# (26,inf) - with the low part INCLUSIVE; high part NOT inclusive; going from 26 to 100

data['my_bins'] = pd.cut(data['values'],bins, labels=['0-10','11-25','>25'], right=False)
data
```
